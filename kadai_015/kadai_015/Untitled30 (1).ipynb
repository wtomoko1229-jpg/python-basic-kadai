```python
class Student:
    # コンストラクタを定義する
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def set_name(self, name):
        self.name = name
    def set_age(self, age):
        self.age = age
    def set_gender(self, gender):
        self.gender = gender
    def exam(self):
      if(self.age >= 75):
        print(f"{self.name}さんは後期高齢者です。")
      else:
        print(f"{self.name}さんは前期高齢者です。")

a = Student("麻生", 85, "男性")
b = Student("高市", 64, "女性")
a.exam()
b.exam()
```
