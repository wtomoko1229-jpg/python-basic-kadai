```python
class Human:
    # コンストラクタを定義する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def set_name(self, name):
        self.name = name
    def set_age(self, age):
        self.age = age
    def printinfo(self):
      print(f"名前は{self.name}です。年齢は{self.age}歳です。")

a = Human("麻生", 85)
b = Human("高市", 64)
a.printinfo()
b.printinfo()
```
