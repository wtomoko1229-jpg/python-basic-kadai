```python
class Fruit:
  def __init__(self,name,price,color):
    self.price = price
    self.name = name
    self.color = color
  def intro(self):
    print(f"{self.name}の値段は{self.price}円で、色は{self.color}です")

fruit1 = Fruit("りんご",100,"red")
fruit2 = Fruit("メロン",2350,"green")
fruit1.intro()
fruit2.intro()
```
