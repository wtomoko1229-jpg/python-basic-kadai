```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}さんは大人です")
      return f"{self.name} is an adult."
    else:
      print(f"{self.name}さんは大人ではありません")
      return f"{self.name} is not an adult."

person1 = Human("mana", 21)
person2 = Human("kokoro", 17)
person3 = Human("suzu", 27)

print(person1.check_adult())
print(person2.check_adult())
print(person3.check_adult())
```
